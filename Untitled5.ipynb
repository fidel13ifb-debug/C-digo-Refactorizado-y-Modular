```python
"""
Módulo de procesamiento y estadística climática.
Refactorizado bajo los estándares de Clean Code y PEP 8.
"""

def limpiar_datos_meteorologicos(datos_crudos):
    """
    Filtra los valores nulos, vacíos y desecha códigos de error o rangos inválidos.
    Responsabilidad única: Depuración de datos.
    """
    datos_limpios = []
    for valor in datos_crudos:
        if valor is not None and valor >= 0:
            datos_limpios.append(float(valor))
    return datos_limpios


def calcular_estadisticas_climaticas(datos_limpios):
    """
    Calcula la media, varianza y desviación estándar de un arreglo limpio.
    Responsabilidad única: Procesamiento matemático.
    """
    total_elementos = len(datos_limpios)

    if total_elementos == 0:
        return None

    media = sum(datos_limpios) / total_elementos
    varianza = sum((x - media) ** 2 for x in datos_limpios) / total_elementos
    desviacion_estandar = varianza ** 0.5

    return {
        "media": media,
        "varianza": varianza,
        "desviacion_estandar": desviacion_estandar
    }


# Ejecución principal del sistema modularizado
if __name__ == "__main__":
    datos_xyz = [12.5, -999, 15.0, None, 18.2, 20.1, -5, 14.8]

    # Paso 1: Limpieza
    datos_depurados = limpiar_datos_meteorologicos(datos_xyz)

    # Paso 2: Cálculo estadístico
    resultado_estadistico = calcular_estadisticas_climaticas(datos_depurados)

    print("Datos depurados:", datos_depurados)
    print("Resultado estadístico:", resultado_estadistico)
```
